```python
CD = loadXFELdata.CrispData(file_bunch_92)
```

```python
def addDrawing(axes=[0, 0.9, 1, 0.23]):
    img = convert_from_path("../../../Drawings/XS1_diag_XTLD.pdf", dpi=300)[0]
    arrow_xy=[922, 172, 30, 135]
    ax_pdf = _plt.gcf().add_axes(axes)
    ax_pdf.imshow(img)
    ax_pdf.axis('off')
    arrow = _patch.Rectangle((arrow_xy[0], arrow_xy[1]), arrow_xy[2], arrow_xy[3], lw=1.8, fill=False, color='r')
    ax_pdf.add_patch(arrow)
    ax_pdf.text(arrow_xy[0]+arrow_xy[2]/2, arrow_xy[1]-9, 'CRISP', color='r', ha='center', va='bottom', fontsize=9)
```

## Plots

```python
plotXFELdata.plotBunchProfileAndCumsum(CD, train=200, percentile=0.95, xlim=[-80, 150], figsize=[11, 5])
addDrawing(axes=[-0.08, 1, 1.1, 0.25])
_plt.tight_layout()
_plt.savefig("{}_Profile_and_cumsum.pdf".format(save_name_prefix), dpi=300, bbox_inches='tight')
```

```python
plotXFELdata.plotLengthAndHistPer(CD, percentile=0.95, bins=50, minCut=5000, maxCut=95000, figsize=[11, 5])
addDrawing(axes=[-0.06, 1, 1.1, 0.25])
_plt.tight_layout()
_plt.savefig("{}_Length_per_train.pdf".format(save_name_prefix), dpi=300, bbox_inches='tight')
```
