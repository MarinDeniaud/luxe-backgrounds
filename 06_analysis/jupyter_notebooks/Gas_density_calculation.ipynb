```python
R = 8.314 # J/K/mol

species_molar_mass_dict = { # g/mol
    "H": 1.008,
    "C": 12.011,
    "O": 16,
    "H2": 2.016,
    "H2O": 18.015,
    "CO": 28.01,
    "N2": 28.014,
    "CO2": 44.01
    
}

def calculateMassDensity(species=["H", "C", "O"],
                         fractions=[0.482, 0.221, 0.297], # no units
                         P=1e-5, # Pa = 100 * mbar
                         T=300, # K
                        ):
    molarMasses = []
    for specie in species:
        molarMasses.append(species_molar_mass_dict[specie])
    totalMolarMass = 0
    for frac, molarMass in zip(fractions, molarMasses):
        totalMolarMass += frac*molarMass
    rho = P*totalMolarMass/(R*T) # g/m^3
    return rho * 1e-6 # g/cm^-3
```
